```python
import numpy as np
import matplotlib.pyplot as plt

#Dane:
kanaly = np.array([801.88, 918.6, 1094.96, 1147.07, 1264.32, 1385.09])
energia_mev = np.array([3.5, 4, 4.76, 5, 5.5, 6])
```

# Punkty pomiarowe na wykresie

```python
plt.figure()
plt.plot(kanaly, energia_mev, 'o')
plt.xlabel("Środkowy kanał")
plt.ylabel("Energia (MeV)")

plt.show()
```

# Dopasowanie prostej

```python
a, b = np.polyfit(kanaly, energia_mev, 1)

x_fit = np.linspace(min(kanaly), max(kanaly), 100)
y_fit = a * x_fit + b

plt.figure()
plt.plot(kanaly, energia_mev, 'o', markersize=6)

plt.plot(x_fit, y_fit, '-', color='red', linewidth=2)

plt.xlabel("Kanał")
plt.ylabel("Energia (MeV)")
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()

print(f"Prosta (y = ax + b): a = {a:.5f} MeV, b = {b:.5f} MeV")
```

## Wykres reszt

### Niepewności reszt

### Wykres

```python
y_model = a * kanaly + b
reszty = energia_mev - y_model

niepewnosci = np.array([0.05, 0.05, 0.05, 0.1, 0.1, 0.1])

plt.errorbar(kanaly, reszty, yerr=niepewnosci, fmt='s', color='green', capsize=5)
plt.axhline(0, color='black')
plt.xlabel('Kanał')
plt.ylabel('Reszty (MeV)')
plt.grid(True, linestyle='--', alpha=0.7)    

plt.show()
```
