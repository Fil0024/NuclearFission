```python
import numpy as np
import matplotlib.pyplot as plt

#Dane:
kanaly = np.array([910,1819.71,2730,3657.55,4550])
energia_mev = np.array([20, 40, 60, 80, 100])
```

## Punkty pomiarowe na wykresie

```python
plt.figure()
plt.plot(kanaly, energia_mev, 'o')
plt.xlabel("Środkowy kanał")
plt.ylabel("Energia (MeV)")

plt.show()
```

## Dopasowanie prostej

```python
a, b = np.polyfit(kanaly, energia_mev, 1)

x_fit = np.linspace(min(kanaly), max(kanaly), 100)
y_fit = a * x_fit + b

plt.figure()
plt.plot(kanaly, energia_mev, 'o', markersize=6)

plt.plot(x_fit, y_fit, '-', color='red', linewidth=2)

plt.xlabel("Kanał")
plt.ylabel("Energia (MeV)")
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()

print(f"Prosta (y = ax + b): a = {a:.5f} MeV, b = {b:.5f} MeV")
```
