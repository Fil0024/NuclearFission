```python
import numpy as np

# --- Stałe fizyczne i parametry eksperymentu ---

# Stała Avogadra [atomy/mol]
NA = 6.022e23 

# Masy molowe [g/mol]
mu_235 = 235.0

# Parametry tarczy z instrukcji
sigma_b = 507.0       # Przekrój czynny w barnach [cite: 73]
sigma = sigma_b * 1e-24 # Przekształcenie na cm^2 (1 b = 10^-24 cm^2)

abundance_235 = 0.10  # Zawartość U-235 w tarczy (10%) [cite: 74]

surf_density_ug = 200.0 # Gęstość powierzchniowa w ug/cm^2 [cite: 75]
surf_density = surf_density_ug * 1e-6 # Gęstość w g/cm^2

radius_mm = 5.0       # Promień tarczy w mm [cite: 75]
radius_cm = radius_mm / 10.0 # Promień w cm

efficiency = 0.32     # Wydajność detekcji (32%) [cite: 76]

print(f"Przekrój czynny: {sigma:.2e} cm^2")
```

### Krok 1: Obliczenie liczby jąder U-235

Najpierw obliczamy powierzchnię tarczy ($S$), a następnie całkowitą masę uranu ($m_{tot}$).
Znając masę, wyznaczamy liczbę jąder U-235 korzystając z zawartości procentowej i stałej Avogadra:

$$N_{U235} = \frac{m_{tot} \cdot \text{abundance} \cdot N_A}{\mu_{235}}$$


```python
# 1. Powierzchnia tarczy [cm^2]
area = np.pi * (radius_cm**2)

# 2. Całkowita masa uranu w tarczy [g]
# Masa = gęstość powierzchniowa * powierzchnia
mass_total = surf_density * area

# 3. Masa samego izotopu U-235 [g]
mass_u235 = mass_total * abundance_235

# 4. Liczba jąder U-235
# N = (masa / masa_molowa) * NA
N_targets = (mass_u235 / mu_235) * NA

print(f"Powierzchnia tarczy: {area:.4f} cm^2")
print(f"Całkowita masa uranu: {mass_total:.2e} g")
print(f"Liczba jąder U-235 (N_targets): {N_targets:.2e}")
```

### Krok 2: Wyznaczenie strumienia neutronów

Związek między obserwowaną częstością zliczeń ($R_{det} = \frac{N_{zliczen}}{t}$) a strumieniem neutronów $\Phi$ wyraża się wzorem:

$$R_{det} = \Phi \cdot \sigma \cdot N_{U235} \cdot \varepsilon$$

Przekształcając wzór, otrzymujemy formułę na strumień:

$$\Phi = \frac{N_{zliczen}}{t \cdot \sigma \cdot N_{U235} \cdot \varepsilon}$$

Gdzie:
* $N_{zliczen}$ - suma zliczeń pod oboma garbami widma rozszczepieniowego.
* $t$ - czas pomiaru w sekundach.

```python
# --- TU WPISZ SWOJE DANE Z POPRZEDNIEJ ANALIZY ---

measured_counts = 4003  # Przykładowa wartość (suma zliczeń fragmentów lekkich i ciężkich)
measurement_time = 176561   # Przykładowy czas pomiaru w sekundach

# -------------------------------------------------

# Częstość zliczeń (Counts Per Second - CPS)
cps = measured_counts / measurement_time

# Obliczenie strumienia
# Phi = CPS / (sigma * N * efficiency)
flux = cps / (sigma * N_targets * efficiency)

print(f"Liczba zliczeń: {measured_counts}")
print(f"Czas pomiaru: {measurement_time} s")
print(f"Częstość zliczeń: {cps:.2f} cps")
print("-" * 30)
print(f"Szacowany strumień neutronów: {flux:.2f} n / (cm^2 * s)")
```
